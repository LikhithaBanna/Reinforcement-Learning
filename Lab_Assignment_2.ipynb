```python
import random
M = int(input("Enter number of machines: "))
machines = {}
rewards = {}
Q = {}
for i in range(M):
    name = input(f"\nEnter name of machine {i + 1}: ")
    reward_input = input("Enter rewards separated by spaces: ")
    reward_list = list(map(float, reward_input.split()))
    machines[i] = name
    rewards[name] = reward_list
    Q[name] = sum(reward_list) / len(reward_list)
print("\n--- Estimated Q Values ---")
for i in range(M):
    name = machines[i]
    print(f"Machine {name}")
    print("Rewards:", rewards[name])
    print(f"Q({name}) = {Q[name]:.2f}")
best_machine = max(Q, key=Q.get)
print(f"\nGreedy Selected Machine: {best_machine}")
```
